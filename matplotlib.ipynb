```python
a =[1,2,3,4]
labels =['python','java','c++','c']
plt.pie(a, labels=labels ,explode=[0.1,0.1,0.1,0.1] , shadow=True)   # explode : pie chart ke slice ko thoda alag karta hai
plt.show()
```

```python
a =[1,2,3,4]
labels =['python','java','c++','c']
c=['black', 'green', 'blue', 'yellow']
plt.pie(a, labels=labels ,explode=[0.1,0.1,0.1,0.1] , shadow=True, colors=c)   # explode : pie chart ke slice ko thoda alag karta hai
plt.show()
```

```python
a =[1,2,3,4]
labels =['python','java','c++','c']
c=['black', 'green', 'blue', 'yellow']
plt.pie(a, labels=labels ,explode=[0.1,0.1,0.1,0.1] , shadow=True, colors=c , autopct= '%1.1f%%')   # explode : pie chart ke slice ko thoda alag karta hai
plt.show()
```

```python
a =[1,2,3,4]
labels =['python','java','c++','c']
c=['black', 'green', 'blue', 'yellow']
plt.pie(a, labels=labels ,explode=[0.1,0.1,0.1,0.1] , shadow=True, colors=c , autopct= '%1.1f%%' ,)   # explode : pie chart ke slice ko thoda alag karta hai
plt.legend()
plt.show()
```
