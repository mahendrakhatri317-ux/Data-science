```python

# 1. Create a class Employee with attributes emp_name, emp_id, and salary. Create 3 objects and display their details.

# output :
class Employee:
    
    def __init__(self, emp_name, emp_id, salary):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.salary = salary

    def display_details(self):
        print(f"ID: {self.emp_id} | Name: {self.emp_name} | Salary: ₹{self.salary}")


emp1 = Employee("Aarav", 101, 50000)
emp2 = Employee("Bhavya", 102, 65000)
emp3 = Employee("Chirag", 103, 55000)


emp1.display_details()
emp2.display_details()
emp3.display_details()
```

```python
# 2. Create a class Calculator with methods for addition, subtraction, multiplication, and division.
# output :

class Calculator:
    def addition(self, a, b):
        return a + b

    def subtraction(self, a, b):
        return a - b

    def multiplication(self, a, b):
        return a * b

    def division(self, a, b):
        if b == 0:
            return "Error: Cannot divide by zero."
        return a / b


calc = Calculator()

print(f"Addition (10 + 5): {calc.addition(10, 5)}")
print(f"Subtraction (10 - 5): {calc.subtraction(10, 5)}")
print(f"Multiplication (10 * 5): {calc.multiplication(10, 5)}")
print(f"Division (10 / 5): {calc.division(10, 5)}")
```

```python
# 3. Create a class Book with a constructor to initialize title, author, and price, then print details.
#output :

class Book:
    
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

   
    def print_details(self):
        print(f"Book Title: '{self.title}'")
        print(f"Author: {self.author}")
        print(f"Price: ₹{self.price}")


my_book = Book("Atomic Habits", "James Clear", 399)
my_book.print_details()
```

```python
# 4. Create a class Vehicle with method display(). Inherit it into class Car and call parent method.
# output :

class Vehicle:
    def display(self):
        print("This is a Vehicle.")


class Car(Vehicle):
    def show_car_details(self):
        print("This is specifically a Car.")
        # Calling the parent method using super()
        super().display()


my_car = Car()
my_car.show_car_details()
```

```python
# 5. Write a Python program using regex to find all digits from the string: "Python123Assignment456"
# output :

import re

text = "Python123Assignment456"


digits = re.findall(r'\d+', text)

print(f"Original String: {text}")
print(f"Extracted Digits: {digits}")
```
