# <span style=color:green>Conditional Statements In Python 😎</span>
**Conditional statements ka use program me decision lene ke liye hota hai.**
Yani agar koi condition True hai to ek code chalega, aur agar False hai to dusra code chalega.

**<span style=color:red>Types of Conditional Statements></span>**

 1.  if<br>
 2.  if-else<br>
 3.  if-elif-else<br>
 4.  Nested if<br>


# <span style= color:blue >if statement 🙄 </span>
if statement ka use condition check karne ke liye hota hai. Agar condition True hoti hai to code execute hota hai.

*<span style= color:pink>syntax 👉</span>*<br><br>
         
     if condition:
        statement

 **Example :-** 


```python
balance = 5000
# withdrawal_amount = int(input("enter the withdrawal amount"))
withdrawal_amount = 1000
if withdrawal_amount <= balance:
    balance -= withdrawal_amount
    print("withdrawal successful")
    print("remaining balance:", balance)
else:
    print("insufficient funds")
```

```python
#### student fail pass program

max=36
if max >= 35:
    print("student passed")
else:
    print("student failed")
```

```python
### trafic light program
light_color = "red"
if light_color == "red":
    print("stop")
elif light_color == "yellow":
    print("get ready")
elif light_color == "green":
    print("go")
```

```python
## voting system
age =18
if age >= 18:
    print("eligible to vote")
else:
    print("not eligible to vote")
```

```python
## dilivery free
order_amount = 1500
if order_amount >= 1000:
    print("free delivery")
else:
    print("delivery charge apply")
```

```python
## student grade program
marks = int(input("Enter the student's marks: "))
if marks >=100:
    print("grade A")
elif marks >=70-75:
    print("grade B")
elif marks >=55-65:
    print("grade C")
elif marks >=35-40:
    print("grade D")
else:
    print(" F")
```

## Nested if statement 👉

```python
## internal exam's ka lafda
attendence = 80
marks = 33
if attendence >= 75:
    if marks >= 36:
        print("pass")
    else:
        print("fail")
else:
    print("low attendence")
```

```python
## transaction from if else to nested if else
pin = int(input("Enter PIN: "))
balance = 12000
if pin == 1234:
    if balance >= 10000:
        print("Transaction successful.")
    else:
        print("Insufficient balance.")
else:
    print("Invalid PIN.")
```
