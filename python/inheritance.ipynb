# Inheritance ✔
* Inheritance is a feature of OOP (Object-Oriented Programming) where one class can use the properties and methods of another class.

* Parent Class / Base Class → The class whose properties are inherited.
* Child Class / Derived Class → The class that inherits properties.

        Inheritance helps in:

* Code reusability
* Reducing duplicate code
* Easy maintenance


Types of Inheritance in Python


| Type                     | Meaning                                    |
| ------------------------ | ------------------------------------------ |
| Single Inheritance       | One parent → One child                     |
| Multiple Inheritance     | Multiple parents → One child               |
| Multilevel Inheritance   | Grandparent → Parent → Child               |
| Hierarchical Inheritance | One parent → Multiple children             |
| Hybrid Inheritance       | Combination of different inheritance types |

```python
# single inheritance

class college:
    pass
class student(college):
    pass
```

```python
class college:
    college_name="PU"
    college_address = "sitapura"
    college_sp=42
    def show(self):
        print(self.college_name)
        print(self.college_address)
        print(self.college_sp)

        
```

```python
# multilevel inheritane

class Person:
    def person_info(self):
        print("This is a person")

class Teacher(Person):
    def teacher_info(self):
        print("This is a teacher")

class HOD(Teacher):
    def hod_info(self):
        print("This is a HOD")

h = HOD()
print(h.person_info())
print(h.teacher_info())
print(h.hod_info()) 

```

```python
# multiple inheritance
class developer:
    def codding(self):
        print("bhaiya ye code likhiye")

class manager:
    def metting(self):
        print("bhaiya ye meeting karni hai")

class team_leader(developer,manager):
    def project(self):
        print("bhaiya ye project complete karna hai")

team= team_leader()
team.codding()                
```

```python
class Company1:
    def company1_info(self):
        print("Company 1: upflairs")


class Company2:
    def company2_info(self):
        print("Company 2: CIT")


class Employee(Company1, Company2):
    def employee_info(self):
        print("Employee Name: Mahendra")


a = Employee()


a.company1_info()
a.company2_info()
a.employee_info()
```
