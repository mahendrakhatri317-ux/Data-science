     ### <span style=color:gray> [1 .] For Loop </span><br>
     for loop ka use sequence (list, tuple, string, range) par iteration karne ke liye hota hai.


```python
name = "MAHEDNRA"
for x in name:
    print(x)
```

```python
name = [8,7,6,5,6,4,3,32,5,6,7,8,]
for y in name:
    print(y)
```

```python
for x in range(1,5):
    print(x)
```

```python
for y in range (5,50,5):
    print(y)
```

 ## [2.] while Loop
 
 Used when the loop should continue until a condition becomes false.

 Example :-

```python
x = 1

while x <= 5:
    print(x)
    x += 1
```

 ## Loop Control Statements 👉

 #### * Break statement 
 Stops the loop immediately.

 *example:-*

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

#### * Continue statement
Skips the current iteration.

*example*

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```
