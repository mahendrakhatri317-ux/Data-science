# Seaborn 

Seaborn Python ki ek advanced Data Visualization Library hai jo Matplotlib ke upar bani hui hai.

**Seaborn is a Python statistical data visualization library built on top of Matplotlib. It provides a high-level interface for creating attractive, informative, and statistical graphics with less code. It is widely used in Data Science, Machine Learning, and Exploratory Data Analysis (EDA).**

## Most Important Seaborn Functions 👉
| Function          | Work                  |
| ----------------- | --------------------- |
| sns.scatterplot() | Scatter Plot          |
| sns.lineplot()    | Line Plot             |
| sns.barplot()     | Bar Plot              |
| sns.histplot()    | Histogram             |
| sns.kdeplot()     | Density Curve         |
| sns.boxplot()     | Outliers              |
| sns.violinplot()  | Distribution          |
| sns.countplot()   | Count Categories      |
| sns.pairplot()    | Relationship Analysis |
| sns.heatmap()     | Correlation           |
| sns.jointplot()   | Joint Distribution    |
| sns.regplot()     | Regression            |

```python
import os 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset('tips')
```

```python
data
```

```python
plt.plot(data['total_bill'])
plt.show()
```

```python
sns.lineplot(data['total_bill'])
plt.show()
```

```python
plt.scatter(data['total_bill'],data['tip'])
plt.show()
```

```python
sns.scatterplot(x = 'total_bill' , y='tip' , data=data)
```

```python
plt.hist(data['total_bill'])
```

```python
sns.histplot(data['total_bill'])
```

```python
sns.kdeplot(data['total_bill'])
```

```python
# catplot and relplot

sns.relplot(x='total_bill' , y='tip' , data=data)
```

```python
# catplot and relplot

sns.catplot(x='total_bill' , y='tip' , data=data)
```

```python
# catplot and relplot

sns.relplot(x='total_bill' , y='tip', hue='time' , data=data)
plt.show()
```

```python
# catplot and relplot

sns.relplot(x='total_bill' , y='tip', hue='time' , kind='scatter' , col='smoker', data=data)
plt.show()
```

```python
# catplot and relplot

sns.relplot(x='total_bill' , y='tip', hue='time' , kind='scatter' , col='day', data=data)
plt.show()
```

```python
# catplot and relplot

sns.relplot(x='total_bill' , y='tip', hue='time' , kind='scatter' , col='day', data=data)
plt.show()
```

```python
sns.relplot(x = 'total_bill' , y = 'tip' , data = data)
plt.title("this is a scatter plot")
plt.xlabel("Total bill" , fontsize = 20 , color='red')
plt.ylabel("Tips" , fontsize = 20 , color = 'red' ) 
```
