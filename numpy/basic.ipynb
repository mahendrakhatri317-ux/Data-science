#  Numpy 
NumPy (Numerical Python) एक open-source Python library है जो multi-dimensional arrays और high-speed mathematical operations प्रदान करती है। यह Data Science, Machine Learning, AI और Scientific Computing में व्यापक रूप से उपयोग की जाती है।

NumPy की Features
### 1. Fast Processing

NumPy का core C language में लिखा गया है, इसलिए Python lists से काफी तेज चलता है।

### 2. Less Memory Usage

NumPy arrays Python lists की तुलना में कम memory लेते हैं।

### 3. Multi-Dimensional Arrays
* 1D Array
* 2D Array
* 3D Array
* N-Dimensional Array

### 4. Mathematical Functions
* Addition
* Subtraction
* Multiplication
* Division
* Power
* Square Root
* Logarithm

### 5. Linear Algebra
* Matrix multiplication
* Determinant
* Inverse
* Eigen values

```python
# ==============================
# NumPy Array Operations Example
# ==============================

import numpy as np

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 6, 8])

print("Array 1 :", arr1)
print("Array 2 :", arr2)

print("\n========== OPERATIONS ==========\n")

# Multiplication
print("1️⃣ Multiplication (*)")
print("arr1 * arr2 =", arr1 * arr2)

# Division
print("\n2️⃣ Division (/)")
print("arr1 / arr2 =", arr1 / arr2)

# Addition
print("\n3️⃣ Addition (+)")
print("arr1 + arr2 =", arr1 + arr2)

# Subtraction
print("\n4️⃣ Subtraction (-)")
print("arr1 - arr2 =", arr1 - arr2)

# Positive Sign with Multiplication
print("\n5️⃣ Multiplication with Unary Plus (* +)")
print("arr1 * +arr2 =", arr1 * +arr2)

# Floor Division
print("\n6️⃣ Floor Division (//)")
print("arr1 // +arr2 =", arr1 // +arr2)

print("\n✅ All operations completed successfully!")
```
