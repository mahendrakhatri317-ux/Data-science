* Git = Version Control System (Software)

## Git क्या है?

Git एक Distributed Version Control System (DVCS) है।<br>

इसे Linus Torvalds ने 2005 में बनाया था।<br><br>

Git की मदद से:-

Code history track कर सकते हैं<br>
Old version वापस ला सकते हैं<br>
Team में काम कर सकते हैं<br>
Changes compare कर सकते हैं<br>

### Git Workflow

```python
Working Directory
       ↓
Staging Area
       ↓
Repository
```

#### *Step 1: Repository Initialize*

```python
# folder me jao 
cd Desktop

# repository banao
git init

### output :- Initialized empty Git repository     (ek hidden .git folder ban jayega)
```

#### *Step 2: status check*

```python
git status

# ye batata hai :- 
#                    konsi file new hai 
#                    konsi file modified hai
#                    konsi staged hai
```

#### *Step 3 : Add karo*

```python
git add .
```

#### *Step 4: Commit*

```python
git commit -m "First commit"
```

#### *Step 5: History check*

```python
git log
```

# GitHub kya hai ?
GitHub एक cloud platform है जहाँ Git repositories online store की जाती हैं।<br>

GitHub से:<br>

Code backup रहता है<br>
Team collaboration होती है<br>
Projects share किए जाते हैं<br>
Portfolio बनाया जाता है<br>

GitHub की official website:

*GitHub*


**Repository क्या होती है?**

Repository (Repo) वह जगह है जहाँ project और उसकी history store होती है।

Example:

ATM_Project<br>
│<br>
├── atm.py<br>
├── users.csv<br>
└── README.md<br>

यह पूरा project एक repository हो सकता है।<br><br>

**GitHub पर Repository बनाना 👉**

* GitHub account बनाओ।
* New Repository पर क्लिक करो।
* Repository name दो।
* Create Repository।
<br><br>

**Local Repo को GitHub से Connect करना 😎**

GitHub repository create होने के बाद:

```python
git remote add origin https://github.com/username/project.git
```

```python
# check ::

git remote -v
```

**Push😎**

Local code GitHub पर भेजना

```python
git push
```

**Pull😎**

GitHub से latest code लाना

```python
git pull origin main
```

**Clone😎**

किसी repository को download करना

```python
git clone https://github.com/user/project.git
```
